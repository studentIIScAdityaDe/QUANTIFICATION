````python
# Import necessary libraries
import sys
sys.path.append('..')

from quantification import (
    calculate_mean,
    calculate_median,
    calculate_std,
    calculate_percentiles,
    summarize_data,
    plot_distribution,
    plot_comparison,
    plot_correlation
)

import numpy as np
import pandas as pd
````

## 1. Basic Statistical Analysis

Let's start with some basic statistical calculations.

````python
# Sample data: student test scores
test_scores = [85, 92, 78, 95, 88, 76, 89, 93, 84, 90, 87, 91, 82, 94, 79]

print("Test Scores:", test_scores)
print(f"\nMean: {calculate_mean(test_scores):.2f}")
print(f"Median: {calculate_median(test_scores):.2f}")
print(f"Standard Deviation: {calculate_std(test_scores):.2f}")
````

````python
# Calculate percentiles
percentiles = calculate_percentiles(test_scores, [10, 25, 50, 75, 90])
print("Percentiles:")
for p, value in percentiles.items():
    print(f"  {p}th percentile: {value:.2f}")
````

````python
# Get a complete summary
summary = summarize_data(test_scores)
print("Complete Statistical Summary:")
for key, value in summary.items():
    print(f"  {key:>10s}: {value:.2f}")
````

## 2. Data Visualization

Now let's create some visualizations of our data.

````python
# Plot the distribution of test scores
plot_distribution(test_scores, 
                 title="Distribution of Test Scores",
                 xlabel="Score",
                 ylabel="Frequency",
                 bins=10)
````

````python
# Compare average scores across different subjects
subject_scores = {
    'Math': 87.5,
    'Science': 91.2,
    'English': 84.3,
    'History': 88.7,
    'Art': 92.8
}

plot_comparison(subject_scores,
               title="Average Scores by Subject",
               xlabel="Subject",
               ylabel="Average Score")
````

## 3. Correlation Analysis

Let's explore the relationship between two variables.

````python
# Example: Study hours vs test scores
study_hours = [2, 3, 1, 5, 4, 2, 4, 5, 3, 4, 3, 5, 2, 5, 2]
scores = [78, 85, 72, 95, 90, 76, 89, 93, 84, 88, 82, 94, 79, 91, 80]

plot_correlation(study_hours, scores,
                title="Study Hours vs Test Scores",
                xlabel="Study Hours",
                ylabel="Test Score")
````

## 4. Working with Larger Datasets

Let's generate and analyze a larger synthetic dataset.

````python
# Generate random data following a normal distribution
np.random.seed(42)  # For reproducibility
large_dataset = np.random.normal(loc=100, scale=15, size=1000)

print(f"Generated {len(large_dataset)} samples")
print("\nStatistical Summary:")
summary = summarize_data(large_dataset)
for key, value in summary.items():
    print(f"  {key:>10s}: {value:.2f}")
````

````python
# Visualize the distribution
plot_distribution(large_dataset,
                 title="Large Dataset Distribution (N=1000)",
                 xlabel="Value",
                 ylabel="Frequency",
                 bins=40)
````

## 5. Practical Example: Real-World Data Analysis

Let's simulate a real-world scenario where we're analyzing measurement data.

````python
# Simulate measurement data from different instruments
instrument_A = np.random.normal(50, 2, 100)
instrument_B = np.random.normal(50.5, 1.5, 100)
instrument_C = np.random.normal(49.8, 2.5, 100)

print("Measurement Comparison:")
print("\nInstrument A:")
for key, value in summarize_data(instrument_A).items():
    print(f"  {key:>10s}: {value:.3f}")

print("\nInstrument B:")
for key, value in summarize_data(instrument_B).items():
    print(f"  {key:>10s}: {value:.3f}")

print("\nInstrument C:")
for key, value in summarize_data(instrument_C).items():
    print(f"  {key:>10s}: {value:.3f}")
````

````python
# Compare mean values
comparison = {
    'Instrument A': calculate_mean(instrument_A),
    'Instrument B': calculate_mean(instrument_B),
    'Instrument C': calculate_mean(instrument_C)
}

plot_comparison(comparison,
               title="Mean Measurements by Instrument",
               xlabel="Instrument",
               ylabel="Mean Measurement")
````
